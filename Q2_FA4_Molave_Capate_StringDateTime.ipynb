```python
# Version 2

first_name = input("Enter first name: ")
last_name = input("Enter last name: ")
section = input("Enter section: ")
favorite_subject = input("Enter favorite subject: ")

full_name = (first_name + " " + last_name).title()

print("\nFormatted Output:")
print("Name:", full_name)
print("Capital Full Name:", full_name.upper())
print("Section:", section.upper())
print("Favorite Subject:", favorite_subject.title())


# I added the capital version of full name.
```

```python
# Version 1

event_title = input("Enter event title: ")
year = input("Enter year: ")
month = input("Enter month: ")
day = input("Enter day: ")

print("\nEvent Details:")
print("Event:", event_title)
print("Year:", year)
print("Month:", month)
print("Day:", day)

# Made basic event deadline input and output.
```

```python
# Version 2

from datetime import date

event_title = input("Enter event title: ")
year = int(input("Enter year: "))
month = int(input("Enter month: "))
day = int(input("Enter day: "))

event_date = date(year, month, day)

print("\nFormatted Output:")
print("Event:", event_title.upper())
print("Date:", f"{event_date.strftime('%B')} {event_date.day}, {event_date.year}")
print("Year:", event_date.year)
print("Month:", event_date.strftime("%B"))
print("Day:", event_date.day)
print("Weekday:", event_date.strftime("%A"))

# Added event deadline using datetime.
```
